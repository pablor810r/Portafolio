# Clasificador de solicitudes artísticas

Modelo simple de Machine Learning para orientar solicitudes del portafolio PABLOR/BLOR. Clasifica textos en tres categorías: `mural`, `digital` y `taller`.

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score, ConfusionMatrixDisplay

dataset = pd.DataFrame({
    'solicitud': [
        'mural para la fachada de una empresa', 'pintura urbana para un espacio publico',
        'intervencion de gran formato en Medellin', 'diseno de mural con identidad de marca',
        'ilustracion digital personalizada', 'pieza digital para redes sociales',
        'obra digital coleccionable NFT', 'animacion para una campana artistica',
        'aprender tecnicas de muralismo', 'taller de arte para estudiantes',
        'charla de arte y tecnologia para universidad', 'curso de aerosol y composicion urbana'
    ],
    'categoria': ['mural'] * 4 + ['digital'] * 4 + ['taller'] * 4
})
dataset
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    dataset['solicitud'], dataset['categoria'], test_size=0.25, random_state=42, stratify=dataset['categoria']
)
modelo = Pipeline([
    ('vectorizador', TfidfVectorizer()),
    ('clasificador', KNeighborsClassifier(n_neighbors=3))
])
modelo.fit(X_train, y_train)
predicciones = modelo.predict(X_test)
print(f'Exactitud de prueba: {accuracy_score(y_test, predicciones):.0%}')
ConfusionMatrixDisplay.from_predictions(y_test, predicciones, cmap='Oranges')
plt.title('Matriz de confusion')
plt.tight_layout()
plt.show()
```

```python
nueva_solicitud = ['Necesito un mural para la entrada de mi negocio']
print('Solicitud:', nueva_solicitud[0])
print('Prediccion:', modelo.predict(nueva_solicitud)[0])
```

## Interpretacion
El dataset es pequeno y demostrativo, por lo que el resultado sirve para orientar el primer contacto y no para tomar decisiones autonomas. En produccion se ampliaria con solicitudes reales anonimizadas y validacion cruzada.
